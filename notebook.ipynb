**and define the argument parameters.**

**Afterwards, let's set or load the input data we want to analyze.**

```python
X = ""
```

**With the parameters and input data set, let's run the whole pipeline...**

```python
X_ = nlp.fit_transform(X, **fit_params); X_
```

**...or execute it step by step, showing the time it takes to complete each stage.**

```python
from time import time

def test(X, named_steps=[]):
    X_ = []
    t0 = time()
    nlp = NLP()

    for i, step in enumerate(named_steps or nlp.named_steps):
        t = time()
        print(f"\nstep {i+1}: '{step}'\ninput: {type(X_[-1] if X_ else X)}")

        # %time \
        X_.append(
            nlp.named_steps[step]
            .fit(X_[-1] if X_ else X, **{k.split("_")[-1]: v for k, v in fit_params.items() if k.startswith(step)})
            .transform(X_[-1] if X_ else X)
        )

        print(f"output: {type(X_[-1] if X_ else X)}\ntime: {float(time()-t):.3f}s")

    print(f"\nFinished execution in {float(time()-t0):.3f}s.")
    return {step: X_[i] for i, step in enumerate(named_steps or nlp.named_steps)}

X_dict = test(X)  # named_steps=["pandas", "tokenizer", "nltk", "spacy", "processor"]
```
